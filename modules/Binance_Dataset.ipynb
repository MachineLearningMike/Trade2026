```python
import os
import sys
import numpy as np
from pathlib import Path
from datetime import datetime, timedelta, timezone


from Binance_file_api_demiss import get_historical_klines
```

```python
save_folder = Path("d:\\trade_data\\history")
start = datetime(2018, 1, 1)
end = datetime.now()
symbol = 'BTCUSDT'
interval = '1m'
data_dir = 'd:/binance_data'
step = 10   # minutes

start_utc = start - timedelta(minutes=step)
while start_utc < end:
    start_utc = start_utc + timedelta(minutes=step)
    end_utc = start_utc + timedelta(minutes=2*24*60)
    
    folder_path = save_folder / str(start_utc.year).zfill(4) / str(start_utc.month).zfill(2) / str(start_utc.day).zfill(2)
    folder_path.mkdir(parents=True, exist_ok=True)
    date_anchor = datetime(start_utc.year, start_utc.month, start_utc.day)
    minutes = int((start_utc - date_anchor).total_seconds() / 60)
    file_path_means = folder_path / (str(minutes).zfill(4) + "_means.npy")    
    file_path_volumes = folder_path / (str(minutes).zfill(4) + "_volumes.npy")

    exist = False
    try:
        arr = np.load(file_path_means)
        arr = np.load(file_path_volumes)
        exist = True
    except:
        pass

    if exist: continue
        
    data, successful = get_historical_klines(symbol=symbol, interval=interval, start_utc=start_utc, end_utc=end_utc, data_dir=data_dir, use_api_fallback=True)
    print(f"{start_utc} {successful}")
    if not successful: continue

    means = (data[:, 1] + data[:, 2]) / 2
    volumes = data[:, 4]

    np.save(file_path_means, means)
    np.save(file_path_volumes, volumes)
    
    
```
